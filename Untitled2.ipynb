```python
# ==========================================================
# ADAPTIVE DRIFT-AWARE PHISHING DETECTION FRAMEWORK
# MTech Research Grade | Single Cell | CPU | Colab Ready
# ==========================================================

print("\n=========== INSTALLING DEPENDENCIES ===========")
!pip -q install river shap tldextract

import os, random, time, math
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
import shap
from river.drift import ADWIN
from collections import deque
from google.colab import drive

DEVICE = "cpu"
torch.manual_seed(42)
np.random.seed(42)

# ==========================================================
# STEP 1: MOUNT DRIVE & LOAD DATA
# ==========================================================

print("\n=========== LOADING DATA ===========")

drive.mount("/content/drive")

DATA_PATH = "/content/drive/MyDrive/data"

train_df = pd.read_csv(f"{DATA_PATH}/train.csv")
test_df  = pd.read_csv(f"{DATA_PATH}/test.csv")
phi_df   = pd.read_csv(f"{DATA_PATH}/phiusiil.csv")

# These global definitions are problematic due to inconsistent column names in dataframes
# URL_COL = "url"
# LABEL_COL = "label"

print("Train:", train_df.shape)
print("Test :", test_df.shape)
print("Phi  :", phi_df.shape)

# ==========================================================
# STEP 2: TOKENIZER
# ==========================================================

class CharTokenizer:
    def __init__(self, max_len=128):
        chars = list("abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789-._~:/?#[]@!$&'()*+,;=%")
        self.char2idx = {c:i+1 for i,c in enumerate(chars)}
        self.char2idx["<PAD>"] = 0
        self.max_len = max_len

    def encode(self, url):
        seq = [self.char2idx.get(c,0) for c in str(url)[:self.max_len]]
        return seq + [0]*(self.max_len-len(seq))

tokenizer = CharTokenizer()
VOCAB_SIZE = len(tokenizer.char2idx) # Dynamically determine vocabulary size
print(f"DEBUG: Tokenizer VOCAB_SIZE calculated as: {VOCAB_SIZE}")

def encode_df(df, url_col_name, label_col_name):
    X = np.array([tokenizer.encode(u) for u in df[url_col_name]])
    y = np.array(df[label_col_name]).reshape(-1,1)
    return torch.tensor(X,dtype=torch.long), torch.tensor(y,dtype=torch.float)

X_train, y_train = encode_df(train_df, url_col_name="url", label_col_name="labels")
X_test,  y_test  = encode_df(test_df,  url_col_name="url", label_col_name="labels")
X_phi,   y_phi   = encode_df(phi_df,   url_col_name="URL", label_col_name="label")

print(f"DEBUG: Maximum index in X_train after encoding: {torch.max(X_train).item()}")

# ==========================================================
# STEP 3: DUAL-LEVEL LIGHTWEIGHT ATTENTION TCN
# ==========================================================

class DualAttentionTCN(nn.Module):
    def __init__(self, vocab=VOCAB_SIZE, embed=64): # Use dynamically calculated VOCAB_SIZE
        super().__init__()
        print(f"DEBUG: DualAttentionTCN initialized with vocab: {vocab}")
        self.embedding = nn.Embedding(vocab, embed, padding_idx=0)
        self.conv1 = nn.Conv1d(embed,128,3,padding=2,dilation=2)
        self.conv2 = nn.Conv1d(128,128,3,padding=4,dilation=4)

        # Feature-level attention
        self.feature_attn = nn.Linear(128,1)

        # Decision-level confidence weighting
        self.score_attn = nn.Linear(128,1)

        self.fc = nn.Linear(128,1)

    def forward(self,x):
        x = self.embedding(x).transpose(1,2)
        h = torch.relu(self.conv1(x))
        h = torch.relu(self.conv2(h))
        h = h.transpose(1,2)

        feat_weights = torch.softmax(self.feature_attn(h), dim=1)
        context = (feat_weights * h).sum(dim=1)

        decision_weight = torch.sigmoid(self.score_attn(context))
        logits = self.fc(context) * decision_weight

        return torch.sigmoid(logits), feat_weights

model = DualAttentionTCN().to(DEVICE)
optimizer = optim.Adam(model.parameters(), lr=0.0005)
criterion = nn.BCELoss()

# ==========================================================
# STEP 4: BASE TRAINING
# ==========================================================

print("\n=========== BASE TRAINING ===========")

for epoch in range(3):
    for i in range(0,len(X_train),128):
        xb = X_train[i:i+128]
        yb = y_train[i:i+128]

        pred,_ = model(xb)
        loss = criterion(pred,yb)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch+1} | Loss {loss.item():.4f}")

# ==========================================================
# STEP 5: ZERO-DAY URL MUTATION ENGINE
# ==========================================================

def mutate_url(url):
    ops = [
        lambda x: x.replace(".", "-"),
        lambda x: "secure-" + x,
        lambda x: x.replace("login","l0gin"),
        lambda x: x.replace("paypal","paypaI"),
        lambda x: x + "/verify-account"
    ]
    return random.choice(ops)(str(url))

# ==========================================================
# STEP 6: DRIFT DETECTION + ONLINE LEARNING
# ==========================================================

print("\n=========== DRIFT-AWARE STREAMING ===========")

drift_detector = ADWIN()
confidence_drift_threshold = 0.3
replay_buffer = deque(maxlen=200)
drift_points = []

model.train()

for i in range(len(X_test)):

    url = test_df.iloc[i]["url"] # Use "url" explicitly for test_df
    label = y_test[i].item()

    if random.random() < 0.3:
        url = mutate_url(url)

    x = torch.tensor([tokenizer.encode(url)],dtype=torch.long)
    y = torch.tensor([[label]],dtype=torch.float)

    start_time = time.time()
    pred,_ = model(x)
    latency = time.time() - start_time

    error = abs(pred.item() - label)
    confidence = abs(pred.item() - 0.5)

    drift_detector.update(error)

    if drift_detector.drift_detected or confidence < confidence_drift_threshold:

        print(f"🚨 Drift at index {i} | Latency {latency:.6f}s")
        drift_points.append(i)

        # Replay + incremental update
        replay_buffer.append((x,y))
        batch = list(replay_buffer)

        for xb,yb in batch:
            p,_ = model(xb)
            l = criterion(p,yb)
            optimizer.zero_grad()
            l.backward()
            optimizer.step()

        # drift_detector.reset() # Removed this line as ADWIN does not have a reset method

# ==========================================================
# STEP 7: DYNAMIC RISK TIER CLASSIFICATION
# ==========================================================

def risk_tier(prob):
    if prob > 0.85:
        return "HIGH RISK"
    elif prob > 0.6:
        return "MEDIUM RISK"
    else:
        return "LOW RISK"

print("\n=========== RISK TIER EXAMPLE ===========")
example = "http://paypal-login-secure.com"
ex_x = torch.tensor([tokenizer.encode(example)],dtype=torch.long)
prob,_ = model(ex_x)
print("Probability:",prob.item())
print("Risk Tier:",risk_tier(prob.item()))

# ==========================================================
# STEP 8: EXPLAINABILITY + STABILITY ANALYSIS
# ==========================================================

print("\n=========== SHAP STABILITY ANALYSIS ===========")

def predict_fn(x):
    with torch.no_grad():
        x = torch.tensor(x,dtype=torch.long)
        return model(x)[0].numpy()

sample_x = np.array([tokenizer.encode(example)])
explainer = shap.Explainer(predict_fn, sample_x)
shap_values = explainer(sample_x)

shap_importance = np.mean(np.abs(shap_values.values))
print("Mean SHAP Magnitude:", shap_importance)

# ==========================================================
# STEP 9: EXTERNAL VALIDATION (PHIUSIIL)
# ==========================================================

print("\n=========== PHIUSIIL VALIDATION ===========")

model.eval()
correct = 0

with torch.no_grad():
    for i in range(len(X_phi)):
        pred,_ = model(X_phi[i:i+1])
        correct += int((pred.item()>0.5)==y_phi[i].item())

print("PhiUSIIL Accuracy:", correct/len(X_phi))

# ==========================================================
# FINAL SUMMARY
# ==========================================================

print("\n=========== FINAL SUMMARY ===========")
print("Drift Events:", len(drift_points))
print("Replay Buffer Size:", len(replay_buffer))
print("System: ADAPTIVE | DRIFT-AWARE | ZERO-DAY ROBUST | EXPLAINABLE")
```
